```python
import numpy as np
```

```python
def detect_pauses_xml(xml_feat):
    xml_data = np.load(xml_feat)
    matrix = xml_data['roll']

    pauses = []

    for col in range(matrix.shape[1]):
        frame = matrix[:,col]

        if not frame.any():
            pauses.append(col)

    return pauses
```

```python
detect_pauses_xml('xml_features/schumann_arabeske.musicxml.npz')
```
